```python
#Obj 3 Find Out What Is Driving Better Sales Performance
#T3.1 Find out which types of products are selling more
top_products = df.groupby('Item Type')['Sales'].sum().sort_values(ascending=False)
print("5 - Top Selling products by Sales: \n",top_products.head(5))
```

```python
#T3.2  Find out which types of outlets are performing better
top_outlet = df.groupby('Outlet Type')['Sales'].sum().sort_values(ascending=False)
print("Top oulet: \n",top_outlet)
```

```python
#T3.3 Understand whether product visibility, weight, ratings, and outlet characteristics are connected with sales.
df.groupby(['Outlet Type']).agg(Total_sales=('Sales','sum'),Average_Item_Weight=('Item Weight','mean'),Average_Rating=('Rating','mean'),Average_visibility=('Item Visibility','mean')).sort_values(by='Total_sales',ascending=False).head(20)
```

```python
#T3.4 Compare better-performing and lower-performing products and outlets to understand the difference.
# Bar charts
top_outlet.plot(kind='bar', title='Sales by Outlet Type', color='teal', rot=45)
plt.ylabel('Total Sales ($)')
plt.show()
```

```python
# 1. Quick Item Type chart
top_products.plot(kind='bar', title='Sales by Item Type', rot=45, grid=True);
```

```python
# Combine both series into a single DataFrame (padding empty spaces with 0)
combined_df = pd.DataFrame({'Item Sales': top_products, 'Outlet Sales': top_outlet})
# Plot them together in one chart
combined_df.plot(kind='bar', figsize=(12, 6), rot=45)

plt.title('Sales Comparison: Item Type vs Outlet Type')
plt.ylabel('Total Sales ($)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#T4.1 : Identify the products that are contributing more to sale
df.groupby(['Item Type'])['Sales'].sum().sort_values(ascending=False).head(5)
```

```python
#T4.2: Identify the outlets and locations that are performing better

performance_Outlet = df.groupby(['Outlet Type','Outlet Location Type']).agg(
    Total_Sales=('Sales','sum')).sort_values(by='Total_Sales',ascending=False)
print("outlets and locations that are performing better", performance_Outlet)
performance_Outlet.plot(kind='bar', figsize=(12, 6), rot=45)
plt.title('Sales Comparison: Outlet Type vs Outlet Location')
plt.ylabel('Total Sales ($)')


```

```python
#T4.3:Identify products, outlets, and areas where performance can be improved
low_performance = df.groupby(['Outlet Type','Outlet Location Type','Item Type']).agg(
    Sales=('Sales','mean')
).sort_values(by='Sales', ascending=False)
print("products, outlets, and areas where performance can be improved",low_performance.tail(10))
```

```python
low_performance.plot(kind='line',figsize=(12, 6), rot=45)
```
